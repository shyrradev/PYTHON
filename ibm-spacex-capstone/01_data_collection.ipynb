```python
import json, urllib.request, pandas as pd
from pathlib import Path
api='https://api.spacexdata.com/v4'
# Each launch links to rockets/{id}, payloads/{id}, launchpads/{id}, cores/{id}.
def get_json(endpoint):
    with urllib.request.urlopen(api+endpoint, timeout=5) as response:
        return json.load(response)
try:
    raw=get_json('/launches/past')
    print('Live launch records:',len(raw))
    print('Example launch:',raw[0]['name'])
except Exception as exc:
    print('Live endpoint unavailable:',type(exc).__name__)
# Fixed published snapshot makes the downstream historical experiment reproducible.
df=pd.read_csv('data/dataset_part_1.csv')
print('Historical Falcon 9 subset:',df.shape)
print('Dates:',df.Date.min(),'to',df.Date.max())
print('Source: https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
print(df.head().to_string(index=False))
```
