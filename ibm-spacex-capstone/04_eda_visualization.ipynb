```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
D=pd.read_csv('data/dataset_part_2.csv');Path('figures').mkdir(exist_ok=True)
for group in ['LaunchSite','Orbit']:
 summary=D.groupby(group).Class.agg(['count','sum','mean']);print(summary.to_string())
 ax=summary['mean'].plot.bar(title='Landing success rate by '+group,ylim=(0,1));ax.set_ylabel('Success rate');plt.tight_layout();plt.savefig('figures/rate_'+group+'.png',dpi=160);plt.close()
summary=D.groupby(D.Date.str[:4]).Class.agg(['count','sum','mean']);print(summary.to_string())
ax=summary['mean'].plot(marker='o',title='Annual landing rate',ylim=(0,1));ax.set_ylabel('Success rate');plt.tight_layout();plt.savefig('figures/annual_rate.png',dpi=160);plt.close()
for x,y in [('FlightNumber','LaunchSite'),('PayloadMass','LaunchSite'),('FlightNumber','Orbit'),('PayloadMass','Orbit')]:
 fig,ax=plt.subplots(figsize=(10,5));ax.scatter(D[x],D[y],c=D.Class,cmap='coolwarm',alpha=.8);ax.set(xlabel=x,ylabel=y,title=x+' vs '+y);plt.tight_layout();plt.savefig('figures/'+x+'_'+y+'.png',dpi=160);plt.close()
print('Saved six scatter/rate charts plus annual trend in figures/.')
```
