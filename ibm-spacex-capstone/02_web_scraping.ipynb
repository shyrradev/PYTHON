```python
from pathlib import Path
import pandas as pd
from io import StringIO
# Retrieved using urllib.request from the public launch history page.
url='https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches'
import urllib.request
source=Path('data/launch_history.html')
html=source.read_text() if source.exists() else urllib.request.urlopen(url,timeout=30).read().decode('utf-8')
tables=pd.read_html(StringIO(html),flavor='lxml')
print('URL:',url,'HTML bytes:',len(html.encode()),'Tables:',len(tables))
for i,t in enumerate(tables):
 if len(t)>10:
  print('Table',i,'shape',t.shape,'columns',str(list(t.columns))[:400])
print('Preview of largest extracted table:')
t=max(tables,key=len)
print(t.head(8).to_string(index=False));t.to_csv('data/scraped_launch_table.csv',index=False)
```
