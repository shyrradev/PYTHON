```python
import pandas as pd
D=pd.read_csv('data/dataset_part_1.csv')
print('Input rows:',len(D));print('Missing values:\n',D.isna().sum().to_string())
D['Date']=pd.to_datetime(D['Date'])
D['PayloadMass']=D['PayloadMass'].fillna(D['PayloadMass'].mean())
D['Class']=D['Outcome'].str.startswith('True').astype(int)
expected=pd.read_csv('data/dataset_part_2.csv')
assert D['Class'].equals(expected['Class'])
print('Outcome counts:\n',D.Outcome.value_counts().to_string())
print('Class counts:\n',D.Class.value_counts().sort_index().to_string())
print('Success rate:',D.Class.mean());print('Encoded model matrix:',pd.read_csv('data/dataset_part_3.csv').shape)
```
