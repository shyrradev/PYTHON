# Interactive map using Folium
IBM Applied Data Science Capstone. Executed code and outputs.

```python
from pathlib import Path
import pandas as pd,folium
from folium.plugins import MarkerCluster
D=pd.read_csv('data/spacex_launch_geo.csv');m=folium.Map([32,-99],zoom_start=4)
cluster=MarkerCluster().add_to(m)
for _,r in D.iterrows():
 folium.CircleMarker([r.Lat,r.Long],radius=6,color='green' if r['class']==1 else 'red',fill=True,popup=str(r['Launch Site'])+' flight '+str(r['Flight Number'])).add_to(cluster)
m.save('launch_map.html')
print(D.groupby('Launch Site')['class'].agg(['sum','count','mean']).to_string())
print('Generated launch_map.html with',len(D),'outcome markers.')
```
